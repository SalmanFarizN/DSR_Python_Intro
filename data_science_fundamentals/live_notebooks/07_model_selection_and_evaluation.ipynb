## 2. Progressive Model Development Hierarchy

Never jump straight to fine-tuning a complex neural network or gradient booster on day one!

1. **Level 1: Lazy Estimator (Dummy Benchmark)**:
   - Always predicts the majority class (`strategy='most_frequent'`).
   - Sets the absolute baseline floor: if your ML model cannot convincingly outperform the dummy baseline, your data lacks predictive signal.
2. **Level 2: Baseline Model (Linear / Logistic)**:
   - Quick to train, zero hyperparameter tuning needed, provides baseline feature coefficients.
3. **Level 3: Advanced Model (Ensemble / Non-Linear)**:
   - Random Forest or Gradient Boosting. Captures non-linear thresholds and multi-way interactions.

### Definitions:
- **True Positive (TP)**: Correctly identified churners.
- **False Positive (FP)**: False alarm (retained customer predicted as churner).
- **False Negative (FN)**: Missed churner (customer walks away unnoticed).
- **True Negative (TN)**: Correctly identified loyal customer.

$$\text{Precision} = \frac{\text{TP}}{\text{TP} + \text{FP}} \quad (\text{Of all predicted churners, how many actually churned?})$$
$$\text{Recall (Sensitivity)} = \frac{\text{TP}}{\text{TP} + \text{FN}} \quad (\text{Of all actual churners, what fraction did we catch?})$$
$$F_1 = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$
