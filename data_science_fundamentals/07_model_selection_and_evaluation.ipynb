```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_curve,
    precision_recall_curve,
    classification_report,
)
from sklearn.model_selection import StratifiedKFold, cross_val_score

import warnings

warnings.filterwarnings("ignore")

plt.style.use(
    "seaborn-v0_8-whitegrid"
    if "seaborn-v0_8-whitegrid" in plt.style.available
    else "default"
)

# Load selected feature matrices prepared in Notebook 05
X_train = pd.read_csv("data/telco_X_train_selected.csv")
X_test = pd.read_csv("data/telco_X_test_selected.csv")
y_train = (pd.read_csv("data/telco_y_train.csv")["Churn"] == "Yes").astype(int)
y_test = (pd.read_csv("data/telco_y_test.csv")["Churn"] == "Yes").astype(int)

print(f"Loaded Modeling Matrices: Train {X_train.shape}, Test {X_test.shape}")
print(f"Features: {list(X_train.columns)}")
```

## 2. Progressive Model Development Hierarchy

Never jump straight to fine-tuning a complex neural network or gradient booster on day one!

<div align="center">
  <img src="./images/model_progression.jpg" width="650" alt="Model Progression" />
  <p><em>Figure 3: The Three Tiers of Progressive Model Development.</em></p>
</div>

1. **Level 1: Lazy Estimator (Dummy Benchmark)**:
   - Always predicts the majority class (`strategy='most_frequent'`).
   - Sets the absolute baseline floor: if your ML model cannot convincingly outperform the dummy baseline, your data lacks predictive signal.
2. **Level 2: Baseline Model (Linear / Logistic)**:
   - Quick to train, zero hyperparameter tuning needed, provides baseline feature coefficients.
3. **Level 3: Advanced Model (Ensemble / Non-Linear)**:
   - Random Forest or Gradient Boosting. Captures non-linear thresholds and multi-way interactions.

<div align="center">
  <img src="./images/mvp_iteration_kniberg.png" width="650" alt="Henrik Kniberg MVP" />
  <p><em>Figure 4: Henrik Kniberg's Agile MVP Principle: Build an end-to-end usable skateboard first, not an isolated wheel. Source: Henrik Kniberg, <a href="https://blog.crisp.se/2016/01/25/henrikkniberg/making-sense-of-mvp">Making Sense of MVP</a>.</em></p>
</div>


```python
# Level 1: Lazy Estimator (Dummy Baseline)
dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(X_train, y_train)

y_pred_dummy = dummy.predict(X_test)
dummy_acc = accuracy_score(y_test, y_pred_dummy)
dummy_rec = recall_score(y_test, y_pred_dummy, zero_division=0)

print(f"--- Tier 1: Lazy Estimator (Majority Class) ---")
print(f"Accuracy: {dummy_acc*100:.2f}%")
print(f"Recall:   {dummy_rec*100:.2f}% (Finds 0 churners!)")
```

```python
# Level 2: Baseline Model (Logistic Regression)
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

baseline_pipe = Pipeline(
    [
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=1000, random_state=42)),
    ]
)

baseline_pipe.fit(X_train, y_train)

y_pred_base = baseline_pipe.predict(X_test)
y_prob_base = baseline_pipe.predict_proba(X_test)[:, 1]

base_acc = accuracy_score(y_test, y_pred_base)
base_rec = recall_score(y_test, y_pred_base)
base_f1 = f1_score(y_test, y_pred_base)
base_auc = roc_auc_score(y_test, y_prob_base)

print(f"--- Tier 2: Baseline Model (Logistic Regression) ---")
print(f"Accuracy: {base_acc*100:.2f}%")
print(f"Recall:   {base_rec*100:.2f}%")
print(f"F1-Score: {base_f1:.4f}")
print(f"ROC-AUC:  {base_auc:.4f}")
```

```python
# Level 3: Advanced Model (Random Forest Ensemble)
rf_model = RandomForestClassifier(
    n_estimators=150,
    max_depth=6,
    class_weight="balanced",  # adjust for ~3:1 class imbalance
    random_state=42,
)

rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)
y_prob_rf = rf_model.predict_proba(X_test)[:, 1]

rf_acc = accuracy_score(y_test, y_pred_rf)
rf_rec = recall_score(y_test, y_pred_rf)
rf_f1 = f1_score(y_test, y_pred_rf)
rf_auc = roc_auc_score(y_test, y_prob_rf)

print(f"--- Tier 3: Advanced Model (Random Forest - Balanced) ---")
print(f"Accuracy: {rf_acc*100:.2f}%")
print(f"Recall:   {rf_rec*100:.2f}% (Notice dramatic recall improvement!)")
print(f"F1-Score: {rf_f1:.4f}")
print(f"ROC-AUC:  {rf_auc:.4f}")
```

## 3. Model Evaluation: The Confusion Matrix & Metric Selection

<div align="center">
  <img src="./images/confusion_matrix.png" width="450" alt="Confusion Matrix" />
  <p><em>Figure 5: Confusion Matrix Structure (True Positive, False Positive, False Negative, True Negative).</em></p>
</div>

### Definitions:
- **True Positive (TP)**: Correctly identified churners.
- **False Positive (FP)**: False alarm (retained customer predicted as churner).
- **False Negative (FN)**: Missed churner (customer walks away unnoticed).
- **True Negative (TN)**: Correctly identified loyal customer.

$$\text{Precision} = \frac{\text{TP}}{\text{TP} + \text{FP}} \quad (\text{Of all predicted churners, how many actually churned?})$$
$$\text{Recall (Sensitivity)} = \frac{\text{TP}}{\text{TP} + \text{FN}} \quad (\text{Of all actual churners, what fraction did we catch?})$$
$$F_1 = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$


```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

cm_base = confusion_matrix(y_test, y_pred_base)
cm_rf = confusion_matrix(y_test, y_pred_rf)

disp1 = ConfusionMatrixDisplay(cm_base, display_labels=["Retained", "Churn"])
disp1.plot(ax=axes[0], cmap="Blues", colorbar=False)
axes[0].set_title(
    f"Baseline Logistic Regression\n(Recall = {base_rec*100:.1f}%)", fontweight="bold"
)

disp2 = ConfusionMatrixDisplay(cm_rf, display_labels=["Retained", "Churn"])
disp2.plot(ax=axes[1], cmap="Greens", colorbar=False)
axes[1].set_title(
    f"Balanced Random Forest\n(Recall = {rf_rec*100:.1f}%)", fontweight="bold"
)

plt.tight_layout()
plt.show()
```

## 4. Threshold Tuning: ROC vs Precision-Recall Curves

By default, binary classifiers use a decision threshold of $p = 0.50$.
However, in customer churn, missing a customer (FN) is 10x more costly than giving an unnecessary discount (FP). By tuning the decision threshold (e.g. classifying anyone with $p > 0.35$ as churn), we catch substantially more churners!


```python
fpr_base, tpr_base, _ = roc_curve(y_test, y_prob_base)
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_prob_rf)

prec_base, rec_base, _ = precision_recall_curve(y_test, y_prob_base)
prec_rf, rec_rf, _ = precision_recall_curve(y_test, y_prob_rf)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 1. ROC Curve
axes[0].plot(
    fpr_base,
    tpr_base,
    label=f"Logistic Regression (AUC = {base_auc:.3f})",
    color="steelblue",
    lw=2,
)
axes[0].plot(
    fpr_rf,
    tpr_rf,
    label=f"Random Forest (AUC = {rf_auc:.3f})",
    color="forestgreen",
    lw=2,
)
axes[0].plot([0, 1], [0, 1], "k--", alpha=0.5)
axes[0].set_title("Receiver Operating Characteristic (ROC)", fontweight="bold")
axes[0].set_xlabel("False Positive Rate (1 - Specificity)")
axes[0].set_ylabel("True Positive Rate (Recall)")
axes[0].legend()

# 2. Precision-Recall Curve
axes[1].plot(rec_base, prec_base, label="Logistic Regression", color="steelblue", lw=2)
axes[1].plot(rec_rf, prec_rf, label="Random Forest", color="forestgreen", lw=2)
axes[1].set_title("Precision-Recall Curve", fontweight="bold")
axes[1].set_xlabel("Recall")
axes[1].set_ylabel("Precision")
axes[1].legend()

plt.tight_layout()
plt.show()
```
